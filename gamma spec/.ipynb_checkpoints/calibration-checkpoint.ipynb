Before running this file, the peak_analysis file should be run.

There are two "base" camibration curves, both created on feb 24 which use a broader dataset for comparison. There are two of these base curves, one at preamp 100 and one at preamp 0.

From these base curves, day-by-day adjustments in error are made.

```python
# -----------------------------
# IMPORTS
# -----------------------------

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import odr

plt.style.use("seaborn-v0_8-darkgrid")
```

```python
# -----------------------------
# LOAD PEAK FIT RESULTS
# -----------------------------

csv_path = Path("peak_analysis_outputs/manual_peak_fits_all.csv")

df = pd.read_csv(csv_path)

print("Rows loaded:", len(df))
display(df.head())
```

```python
# -----------------------------
# SELECT FEB 24 CALIBRATION DATA ("base curve")
# -----------------------------

feb24 = df[df["file"].str.contains("calibration/feb_24")].copy()

# remove unlabeled peaks
feb24 = feb24.dropna(subset=["theoretical_energy_keV"])

print("Feb 24 calibration rows:", len(feb24))

display(feb24[[
    "file",
    "center_ADC",
    "center_ADC_err",
    "theoretical_energy_keV",
    "theoretical_energy_keV_err"
]].head())
```

```python
# -----------------------------
# SPLIT PREAMP SETTINGS
# -----------------------------

preamp0 = feb24[feb24["file"].str.contains("preamp_0")]
preamp100 = feb24[feb24["file"].str.contains("preamp_100")]

print("preamp0 rows:", len(preamp0))
print("preamp100 rows:", len(preamp100))
```

```python
# -----------------------------
# ODR CALIBRATION FIT
# -----------------------------
'''
we have x and y errors, so I use orthogonal distance regression (ODR) to propagate ADC center error and theoretical energy error
'''

def linear_model(B, x):
    """
    Energy = m*(ADC) + b
    """
    m, b = B
    return m * x + b


def fit_calibration(data):

    x = data["center_ADC"].values
    xerr = data["center_ADC_err"].values

    y = data["theoretical_energy_keV"].values
    yerr = data["theoretical_energy_keV_err"].values

    model = odr.Model(linear_model)

    data_odr = odr.RealData(
        x,
        y,
        sx=xerr,
        sy=yerr
    )

    odr_obj = odr.ODR(
        data_odr,
        model,
        beta0=[1.0, 0.0]  # initial guess
    )

    out = odr_obj.run()

    m, b = out.beta
    m_err, b_err = out.sd_beta

    return {
        "m": m,
        "b": b,
        "m_err": m_err,
        "b_err": b_err,
        "odr_output": out
    }
```

```python
# -----------------------------
# FIT BASE CALIBRATION CURVES
# -----------------------------

fit0 = fit_calibration(preamp0)
fit100 = fit_calibration(preamp100)

print("PREAMP 0")
print(f"slope = {fit0['m']:.6f} ± {fit0['m_err']:.6f}")
print(f"offset = {fit0['b']:.6f} ± {fit0['b_err']:.6f}")

print()

print("PREAMP 100")
print(f"slope = {fit100['m']:.6f} ± {fit100['m_err']:.6f}")
print(f"offset = {fit100['b']:.6f} ± {fit100['b_err']:.6f}")
```

```python
# -----------------------------
# PLOT CALIBRATION
# -----------------------------

def plot_calibration(data, fit, title):

    x = data["center_ADC"]
    xerr = data["center_ADC_err"]

    y = data["theoretical_energy_keV"]
    yerr = data["theoretical_energy_keV_err"]

    m = fit["m"]
    b = fit["b"]

    xfit = np.linspace(x.min()*0.9, x.max()*1.1, 500)
    yfit = m*xfit + b

    plt.figure(figsize=(7,5))

    plt.errorbar(
        x,
        y,
        xerr=xerr,
        yerr=yerr,
        fmt="o",
        label="Calibration peaks"
    )

    plt.plot(xfit, yfit, label="ODR fit")

    plt.xlabel("ADC Channel")
    plt.ylabel("Energy (keV)")
    plt.title(title)

    plt.legend()

    plt.show()
```

```python
# -----------------------------
# SHOW FITS
# -----------------------------

plot_calibration(preamp0, fit0, "Base Calibration (Feb 24) — Preamp 0")

plot_calibration(preamp100, fit100, "Base Calibration (Feb 24) — Preamp 100")
```

```python

# -----------------------------
# SAVE BASE CALIBRATION
# -----------------------------

calibration_dir = Path("calibration_data")
calibration_dir.mkdir(parents=True, exist_ok=True)

base_cal = pd.DataFrame([
    {
        "preamp": "preamp0",
        "slope": fit0["m"],
        "slope_err": fit0["m_err"],
        "offset": fit0["b"],
        "offset_err": fit0["b_err"]
    },
    {
        "preamp": "preamp100",
        "slope": fit100["m"],
        "slope_err": fit100["m_err"],
        "offset": fit100["b"],
        "offset_err": fit100["b_err"]
    }
])

base_cal_path = calibration_dir / "base_calibration_feb24.csv"
base_cal.to_csv(base_cal_path, index=False)

print(f"Saved: {base_cal_path}")
display(base_cal)
```


## Daily calibration-error adjustments

The Feb. 24 base calibration is treated as the central calibration curve.  
To account for day-to-day disagreement between that curve and the sparse daily calibration data, we estimate an **additional empirical calibration uncertainty** for each `(day, preamp)` pair.

This is done in **energy space**. For each daily calibration peak, we evaluate the base-curve prediction

\[
E_\text{pred} = m x + b
\]

and compare it to the corresponding reference energy \(E_\text{true}\).  
The formal residual uncertainty for each point is

\[
\sigma_{r,i}^2 = \sigma_{E,\text{pred},i}^2 + \sigma_{E,\text{true},i}^2.
\]

For a given day and preamp, we then estimate an additional empirical uncertainty term \(\tau\) such that the residual scatter beyond the formal uncertainty is absorbed into

\[
\sigma_{\text{total},i}^2 = \sigma_{r,i}^2 + \tau^2.
\]

Here we use the moment estimator

\[
\tau = \sqrt{\max\left(0,\; \langle r^2 \rangle - \langle \sigma_r^2 \rangle\right)}.
\]

This produces a conservative day-specific error adjustment without changing the central calibration curve itself.


```python

# -----------------------------
# PARSE DAILY METADATA
# -----------------------------

import re

def parse_metadata(file_string):
    s = str(file_string).lower()

    day = None
    for tag in ["feb24", "feb25", "feb26", "mar3", "mar5"]:
        if tag in s:
            day = tag
            break

    temp_C = np.nan
    m = re.search(r'_(\d+(?:\.\d+)?)c', s)
    if m:
        temp_C = float(m.group(1))

    preamp = np.nan
    m = re.search(r'preamp[_]?(\d+)', s)
    if m:
        preamp = int(m.group(1))

    isotope = None
    if "co60" in s or "/co_" in s:
        isotope = "Co60"
    elif "cs" in s:
        isotope = "Cs137"
    elif "ba" in s:
        isotope = "Ba"
    elif "na" in s:
        isotope = "Na22"
    elif "co57" in s:
        isotope = "Co57"

    return pd.Series({
        "day": day,
        "temp_C": temp_C,
        "preamp": preamp,
        "isotope": isotope
    })

meta = df["file"].apply(parse_metadata)
df_daily_meta = pd.concat([df.copy(), meta], axis=1)

daily = df_daily_meta[df_daily_meta["file"].str.contains("calibration/daily", na=False)].copy()
daily = daily.dropna(subset=["theoretical_energy_keV", "center_ADC", "center_ADC_err"]).copy()
daily["theoretical_energy_keV_err"] = daily["theoretical_energy_keV_err"].fillna(0.01)

print("Daily calibration rows:", len(daily))
display(daily[[
    "file",
    "day",
    "temp_C",
    "preamp",
    "isotope",
    "center_ADC",
    "center_ADC_err",
    "theoretical_energy_keV",
    "theoretical_energy_keV_err"
]].sort_values(["day", "preamp", "theoretical_energy_keV"]))
```

```python

# -----------------------------
# BASE-CURVE RESIDUALS FOR DAILY CALIBRATION PEAKS
# -----------------------------

def get_base_params(preamp):
    if int(preamp) == 0:
        return fit0["m"], fit0["b"], fit0["m_err"], fit0["b_err"]
    elif int(preamp) == 100:
        return fit100["m"], fit100["b"], fit100["m_err"], fit100["b_err"]
    else:
        raise ValueError(f"Unexpected preamp value: {preamp}")

def energy_from_base(adc, preamp):
    m, b, sm, sb = get_base_params(preamp)
    return m * adc + b

def energy_err_base(adc, adc_err, preamp):
    m, b, sm, sb = get_base_params(preamp)

    dE_dm = adc
    dE_dx = m
    dE_db = 1.0

    var = (dE_dm * sm)**2 + (dE_dx * adc_err)**2 + (dE_db * sb)**2
    return np.sqrt(var)

daily_eval = daily.copy()

daily_eval["E_pred_base"] = daily_eval.apply(
    lambda r: energy_from_base(
        adc=float(r["center_ADC"]),
        preamp=int(r["preamp"])
    ),
    axis=1
)

daily_eval["E_pred_base_err"] = daily_eval.apply(
    lambda r: energy_err_base(
        adc=float(r["center_ADC"]),
        adc_err=float(r["center_ADC_err"]),
        preamp=int(r["preamp"])
    ),
    axis=1
)

daily_eval["resid_base_keV"] = (
    daily_eval["E_pred_base"] - daily_eval["theoretical_energy_keV"]
)

daily_eval["resid_base_err_keV"] = np.sqrt(
    daily_eval["E_pred_base_err"]**2 +
    daily_eval["theoretical_energy_keV_err"]**2
)

daily_eval["pull_base"] = (
    daily_eval["resid_base_keV"] / daily_eval["resid_base_err_keV"]
)

display(daily_eval[[
    "day",
    "preamp",
    "isotope",
    "theoretical_energy_keV",
    "E_pred_base",
    "resid_base_keV",
    "resid_base_err_keV",
    "pull_base"
]].sort_values(["preamp", "day", "theoretical_energy_keV"]))
```

```python

# -----------------------------
# ESTIMATE DAY-BY-DAY EXTRA CALIBRATION ERROR
# -----------------------------

adjustment_rows = []

for (day, preamp), sub in daily_eval.groupby(["day", "preamp"]):
    r = sub["resid_base_keV"].to_numpy(float)
    s = sub["resid_base_err_keV"].to_numpy(float)

    mean_r2 = np.mean(r**2)
    mean_s2 = np.mean(s**2)

    # extra calibration uncertainty in keV needed to explain excess scatter
    extra_sigma_keV = np.sqrt(max(0.0, mean_r2 - mean_s2))

    wrss = np.sum((r / s)**2)
    dof_naive = max(len(sub), 1)
    redchi2_naive = wrss / dof_naive

    adjustment_rows.append({
        "day": str(day),
        "preamp": int(preamp),
        "n_points": int(len(sub)),
        "temp_C_mean": float(sub["temp_C"].mean()) if "temp_C" in sub else np.nan,
        "mean_abs_resid_keV": float(np.mean(np.abs(r))),
        "rms_resid_keV": float(np.sqrt(np.mean(r**2))),
        "mean_formal_err_keV": float(np.sqrt(mean_s2)),
        "extra_sigma_keV": float(extra_sigma_keV),
        "redchi2_base_naive": float(redchi2_naive)
    })

daily_error_adjustments = pd.DataFrame(adjustment_rows).sort_values(["preamp", "day"]).reset_index(drop=True)

display(daily_error_adjustments)
```

```python

# -----------------------------
# SAVE DAILY ERROR ADJUSTMENTS
# -----------------------------

daily_error_path = calibration_dir / "daily_error_adjustments.csv"
daily_error_adjustments.to_csv(daily_error_path, index=False)

print(f"Saved: {daily_error_path}")
display(daily_error_adjustments)
```

```python

# -----------------------------
# OPTIONAL: EXAMPLE OF HOW TO APPLY THE SAVED ERROR ADJUSTMENT
# -----------------------------
# Suppose a later analysis has a formal calibrated-energy uncertainty sigma_E_formal
# for a spectrum taken on a given day and preamp. Then the total uncertainty can be
# inflated by adding the daily calibration term in quadrature:
#
# sigma_E_total = np.sqrt(sigma_E_formal**2 + extra_sigma_keV**2)

# Example lookup table:
daily_error_lookup = {
    (row["day"], int(row["preamp"])): float(row["extra_sigma_keV"])
    for _, row in daily_error_adjustments.iterrows()
}

print("Lookup examples:")
for key, value in daily_error_lookup.items():
    print(key, "->", value, "keV")
```
