This notebook is only for debugging matplotlib interactivity and click capture.

The purpose is to verify that:
1. the widget backend is active,
2. click events are captured,
3. clicked x-values can be stored,
4. this interaction works before integrating it into peak fitting.

```python
# -----------------------------
# MATPLOTLIB BACKEND
# -----------------------------

%matplotlib widget
```

```python
# -----------------------------
# IMPORTS
# -----------------------------

import numpy as np
import matplotlib
import matplotlib.pyplot as plt
```

```python
# -----------------------------
# BACKEND DIAGNOSTIC
# -----------------------------

print("Matplotlib backend:", matplotlib.get_backend())

try:
    import ipympl
    print("ipympl imported successfully.")
    print("ipympl version:", ipympl.__version__)
except Exception as e:
    print("ipympl import failed:", repr(e))
```

```python
# -----------------------------
# SIMPLE TEST PLOT
# -----------------------------

x = np.linspace(0, 10, 500)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Interactive test plot")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.show()
```

```python
# -----------------------------
# CLICK COLLECTOR
# -----------------------------

clicked_points = []

def onclick(event):
    global clicked_points

    if event.inaxes is None:
        return

    clicked_points.append((event.xdata, event.ydata))
    print(f"Clicked: x={event.xdata:.3f}, y={event.ydata:.3f}")
```

```python
# -----------------------------
# ATTACH CLICK HANDLER
# -----------------------------

clicked_points = []

x = np.linspace(0, 10, 500)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Click inside the axes several times")

cid = fig.canvas.mpl_connect("button_press_event", onclick)

plt.show()

print("Click on the plot, then run the next cell.")
```

```python
# -----------------------------
# INSPECT STORED CLICKS
# -----------------------------

print("Clicked points:")
print(clicked_points)
print("Number of clicks:", len(clicked_points))
```
