# Compton feature analysis

This notebook is for **Cs-only** spectra. It fits two manually chosen low-energy features in each spectrum:

1. the **backscatter peak**
2. a **Gaussian proxy for the Compton edge**

The fit centers are recorded both in **ADC** and in **keV**. The energy calibration is taken from:

- `calibration_data/base_calibration_feb24.csv`
- `calibration_data/daily_error_adjustments.csv`

The daily adjustment file is used here as an **extra calibration uncertainty term** rather than as a new day-by-day slope/offset refit. So the central conversion is still the Feb. 24 base calibration, while the quoted energy errors are inflated by the day-specific `extra_sigma_keV` when available.

## Recommended workflow

1. Run the setup cells.
2. Build the Cs file list.
3. Use the preview helpers to inspect a spectrum.
4. Fill in the manual fit windows table for the runs you want.
5. Run the fit/export cells.

This notebook writes its outputs to `compton_analysis_outputs/`.


```python
# -----------------------------
# IMPORTS
# -----------------------------
import re
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.optimize import curve_fit

pd.set_option("display.max_columns", None)
plt.rcParams["figure.figsize"] = (10, 5)
```

```python
# -----------------------------
# FILE LOCATIONS
# -----------------------------
ROOT = Path("processed_data")
BACKGROUND_FILE = ROOT / "calibration" / "background_550PM_feb26_213PM_mar3.csv"

OUTPUT_ROOT = Path("compton_analysis_outputs")
PLOT_DIR = OUTPUT_ROOT / "plots"
PER_RUN_DIR = OUTPUT_ROOT / "per_run_csv"
MASTER_FILE = OUTPUT_ROOT / "compton_feature_fits_all.csv"

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
PLOT_DIR.mkdir(parents=True, exist_ok=True)
PER_RUN_DIR.mkdir(parents=True, exist_ok=True)

def first_existing(*paths):
    for p in paths:
        p = Path(p)
        if p.exists():
            return p
    return Path(paths[0])

BASE_CAL_FILE = first_existing(
    "calibration_data/base_calibration_feb24.csv",
    "base_calibration_feb24.csv",
    "/mnt/data/base_calibration_feb24.csv",
)

DAILY_ERR_FILE = first_existing(
    "calibration_data/daily_error_adjustments.csv",
    "daily_error_adjustments.csv",
    "/mnt/data/daily_error_adjustments.csv",
)

print("ROOT:", ROOT)
print("BACKGROUND_FILE:", BACKGROUND_FILE)
print("BASE_CAL_FILE:", BASE_CAL_FILE)
print("DAILY_ERR_FILE:", DAILY_ERR_FILE)
```

```python
# -----------------------------
# LOAD CALIBRATION TABLES
# -----------------------------
base_cal_df = pd.read_csv(BASE_CAL_FILE)
daily_err_df = pd.read_csv(DAILY_ERR_FILE)

display(base_cal_df)
display(daily_err_df)

def normalize_preamp_value(value):
    text = str(value).lower().replace("_", "").replace(" ", "")
    if "100" in text:
        return 100
    if "0" in text:
        return 0
    try:
        return int(float(value))
    except Exception:
        return np.nan

base_cal_df["preamp_norm"] = base_cal_df["preamp"].map(normalize_preamp_value)
daily_err_df["preamp_norm"] = daily_err_df["preamp"].map(normalize_preamp_value)

BASE_CAL_MAP = {
    int(row["preamp_norm"]): row
    for _, row in base_cal_df.dropna(subset=["preamp_norm"]).iterrows()
}

DAILY_ERR_MAP = {
    (str(row["day"]).lower(), int(row["preamp_norm"])): row
    for _, row in daily_err_df.dropna(subset=["preamp_norm"]).iterrows()
}

print("Base calibration keys:", sorted(BASE_CAL_MAP))
print("Daily error keys:", sorted(DAILY_ERR_MAP))
```

```python
# -----------------------------
# PARSING HELPERS
# -----------------------------
def normalize_relpath(path, root=ROOT):
    path = Path(path)
    try:
        rel = path.relative_to(root)
    except ValueError:
        rel = path
    return rel.as_posix()

def extract_day_from_path(relpath):
    lower = str(relpath).lower()
    if "feb_24" in lower or "feb24" in lower:
        return "feb24"
    if "feb_26" in lower or "feb26" in lower:
        return "feb26"
    if "mar_3" in lower or "mar3" in lower:
        return "mar3"
    if "mar_5" in lower or "mar5" in lower:
        return "mar5"
    return None

def extract_preamp_from_path(relpath):
    lower = str(relpath).lower().replace("_", "")
    if "preamp100" in lower:
        return 100
    if "preamp0" in lower:
        return 0

    # All of the Cs geometry / ordinary runs in this data set appear to be preamp-0.
    if "cs" in lower:
        return 0

    return None

def parse_temperature_from_path(relpath):
    m = re.search(r"([0-9]+(?:\.[0-9]+)?)c", str(relpath).lower())
    return float(m.group(1)) if m else np.nan
```

```python
# -----------------------------
# LOAD SPECTRUM / BACKGROUND SUBTRACTION
# -----------------------------
def _extract_first_number(text):
    if text is None:
        return None
    m = re.search(r"[-+]?\d*\.?\d+", str(text))
    if m is None:
        return None
    try:
        return float(m.group())
    except Exception:
        return None

def load_spectrum(csv_path):
    csv_path = Path(csv_path)

    with open(csv_path, "r", encoding="utf-8", errors="replace") as f:
        lines = f.readlines()

    metadata = {}
    data_start_idx = None

    for i, line in enumerate(lines):
        stripped = line.strip()

        if stripped == "Channel Data:":
            data_start_idx = i + 1
            break

        if stripped.startswith("Elapsed Real Time:"):
            parts = stripped.split(",", 1)
            metadata["Elapsed Real Time"] = parts[1].strip() if len(parts) > 1 else ""
            continue

        if "Elapsed Real Time:" in stripped:
            value = stripped.split("Elapsed Real Time:", 1)[1]
            value = value.split(",", 1)[1].strip() if "," in value else value.strip()
            metadata["Elapsed Real Time"] = value
            continue

        if stripped.startswith("Start Time:"):
            metadata["Start Time"] = stripped.split("Start Time:", 1)[1].lstrip(",").strip()
            continue

        if stripped.startswith("End Time:"):
            metadata["End Time"] = stripped.split("End Time:", 1)[1].lstrip(",").strip()
            continue

        if "," in stripped:
            left, right = stripped.split(",", 1)
            left = left.strip().rstrip(":")
            right = right.strip()
            if left and right and left not in metadata:
                metadata[left] = right

    if data_start_idx is None:
        raise RuntimeError(f"Could not locate 'Channel Data:' block in {csv_path}")

    if data_start_idx < len(lines):
        header_line = lines[data_start_idx].strip().lower()
        if header_line.startswith("channel"):
            data_start_idx += 1

    numeric_rows = []
    for line in lines[data_start_idx:]:
        stripped = line.strip()
        if not stripped:
            continue
        parts = [p.strip() for p in stripped.split(",")]
        if len(parts) < 3:
            continue
        try:
            channel = float(parts[0])
            counts = float(parts[2])
            numeric_rows.append((channel, counts))
        except ValueError:
            continue

    if len(numeric_rows) == 0:
        raise RuntimeError(f"No numeric rows found in {csv_path}")

    arr = np.asarray(numeric_rows, dtype=float)
    x = arr[:, 0]
    counts = arr[:, 1]

    elapsed = _extract_first_number(metadata.get("Elapsed Real Time"))
    metadata["Elapsed Real Time (s)"] = elapsed
    return x, counts, metadata

def counts_to_rate(counts, elapsed_time_seconds):
    if elapsed_time_seconds is None or elapsed_time_seconds <= 0:
        raise RuntimeError("Elapsed real time is missing or nonpositive.")
    return np.asarray(counts, dtype=float) / float(elapsed_time_seconds)

def background_subtracted_rate(sample_relpath, background_path=BACKGROUND_FILE):
    sample_path = ROOT / sample_relpath if not Path(sample_relpath).is_absolute() else Path(sample_relpath)

    x_s, counts_s, meta_s = load_spectrum(sample_path)
    x_b, counts_b, meta_b = load_spectrum(background_path)

    rate_s = counts_to_rate(counts_s, meta_s["Elapsed Real Time (s)"])
    rate_b = counts_to_rate(counts_b, meta_b["Elapsed Real Time (s)"])

    # Robust against grid mismatch: interpolate the background onto the sample grid.
    if len(x_s) != len(x_b) or not np.allclose(x_s, x_b):
        rate_b = np.interp(x_s, x_b, rate_b, left=rate_b[0], right=rate_b[-1])

    adjusted = rate_s - rate_b
    return x_s, rate_s, rate_b, adjusted, meta_s, meta_b
```

```python
# -----------------------------
# ENERGY CALIBRATION HELPERS
# -----------------------------
def adc_to_energy(adc, adc_err, relpath):
    adc = np.asarray(adc, dtype=float)
    adc_err = np.asarray(adc_err, dtype=float)

    day = extract_day_from_path(relpath)
    preamp = extract_preamp_from_path(relpath)

    if preamp not in BASE_CAL_MAP:
        raise RuntimeError(f"No base calibration found for preamp={preamp} from file: {relpath}")

    row = BASE_CAL_MAP[preamp]
    slope = float(row["slope"])
    slope_err = float(row["slope_err"])
    offset = float(row["offset"])
    offset_err = float(row["offset_err"])

    energy = slope * adc + offset

    extra_sigma = 0.0
    if day is not None and (day, preamp) in DAILY_ERR_MAP:
        extra_sigma = float(DAILY_ERR_MAP[(day, preamp)]["extra_sigma_keV"])

    energy_err = np.sqrt(
        (slope * adc_err) ** 2 +
        (adc * slope_err) ** 2 +
        offset_err ** 2 +
        extra_sigma ** 2
    )

    return energy, energy_err, slope, slope_err, offset, offset_err, extra_sigma, day, preamp
```

```python
# -----------------------------
# CS FILE LIST
# -----------------------------
irrelevant = {
    "calibration/TESTFILE_FOR_CONVERSION_BGND.csv",
    "calibration/background_550PM_feb26_213PM_mar3.csv",
}

def get_cs_candidate_files(root=ROOT):
    out = []
    for path in sorted(root.rglob("*.csv")):
        rel = normalize_relpath(path)
        lower = rel.lower()
        if rel in irrelevant:
            continue
        if ".ipynb_checkpoints" in lower:
            continue
        if "cs" not in lower:
            continue
        out.append(rel)
    return out

cs_files = get_cs_candidate_files()
print("Number of Cs candidate files:", len(cs_files))
for i, f in enumerate(cs_files):
    print(f"{i:2d} : {f}")
```

## Manual fit-window table

Edit the table below. Each row is one **file**, with one manually chosen fit window and center guess for:

- `backscatter_peak`
- `compton_edge`

The fit model in both cases is:

\[
A\exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right) + mx + b
\]

The Compton edge is not literally Gaussian; this is just a consistent **proxy fit** so you can compare runs.


```python
# -----------------------------
# MANUAL WINDOW TABLE
# -----------------------------
# Fill this in by hand after previewing the relevant spectra.
# You can start with a subset and append more rows later.

manual_windows = pd.DataFrame([
    # Example row format:
    # {
    #     "file": "data/feb_26/cs137/Cs_22.4C_feb26_run1.csv",
    #     "backscatter_xmin": 120,
    #     "backscatter_xmax": 170,
    #     "backscatter_guess": 145,
    #     "edge_xmin": 320,
    #     "edge_xmax": 360,
    #     "edge_guess": 340,
    # },
])

display(manual_windows)
```

```python
# -----------------------------
# PREVIEW HELPERS
# -----------------------------
def preview_spectrum(file_relpath, xmode="adc", xlim=None, ylim=None, annotate_defaults=True):
    x_adc, raw_rate, bg_rate, adjusted_rate, meta_s, meta_b = background_subtracted_rate(file_relpath)

    adc_err_zeros = np.zeros_like(x_adc, dtype=float)
    x_keV, _, slope, slope_err, offset, offset_err, extra_sigma, day, preamp = adc_to_energy(
        x_adc, adc_err_zeros, file_relpath
    )

    if xmode.lower() == "energy":
        xplot = x_keV
        xlabel = "Energy (keV)"
    else:
        xplot = x_adc
        xlabel = "ADC channel"

    fig, ax = plt.subplots(figsize=(11, 5))
    ax.plot(xplot, adjusted_rate, lw=1.2, label="Background-subtracted rate")
    ax.plot(xplot, raw_rate, alpha=0.5, label="Raw sample rate")
    ax.plot(xplot, bg_rate, alpha=0.5, label="Background rate")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Rate (counts/s)")
    ax.set_title(
        f"{file_relpath}\npreamp={preamp}, day={day}, E = ({slope:.6f}) ADC + ({offset:.6f}), extra_sigma={extra_sigma:.3f} keV"
    )
    ax.grid(True, alpha=0.3)
    ax.legend()

    if xlim is not None:
        ax.set_xlim(*xlim)
    if ylim is not None:
        ax.set_ylim(*ylim)

    if annotate_defaults:
        row = manual_windows[manual_windows["file"] == file_relpath]
        if len(row) == 1:
            row = row.iloc[0]
            for xmin, xmax, guess, color, label in [
                (row["backscatter_xmin"], row["backscatter_xmax"], row["backscatter_guess"], "tab:orange", "backscatter"),
                (row["edge_xmin"], row["edge_xmax"], row["edge_guess"], "tab:green", "edge"),
            ]:
                if np.isfinite(xmin) and np.isfinite(xmax):
                    if xmode.lower() == "energy":
                        xmin = slope * xmin + offset
                        xmax = slope * xmax + offset
                        guess = slope * guess + offset
                    ax.axvspan(xmin, xmax, alpha=0.15, color=color)
                    ax.axvline(guess, ls="--", color=color, lw=1.5, label=label)

    plt.show()

# Example:
# preview_spectrum(cs_files[0], xmode="adc", xlim=(80, 420))
# preview_spectrum(cs_files[0], xmode="energy", xlim=(80, 500))
```

```python
# -----------------------------
# FIT MODELS
# -----------------------------
def gaussian(x, A, mu, sigma):
    sigma = abs(float(sigma))
    return A * np.exp(-((x - mu) ** 2) / (2.0 * sigma ** 2))

def gaussian_plus_linear(x, A, mu, sigma, m, b):
    return gaussian(x, A, mu, sigma) + m * x + b

def fit_feature_window(x, y, xmin, xmax, mu_guess, feature_label, file_relpath):
    mask = (x >= xmin) & (x <= xmax)
    xfit = np.asarray(x[mask], dtype=float)
    yfit = np.asarray(y[mask], dtype=float)

    if len(xfit) < 6:
        raise RuntimeError(f"{file_relpath} / {feature_label}: not enough points in fit window.")

    left_n = max(2, len(xfit) // 8)
    right_n = max(2, len(xfit) // 8)

    x_left = np.median(xfit[:left_n])
    x_right = np.median(xfit[-right_n:])
    y_left = np.median(yfit[:left_n])
    y_right = np.median(yfit[-right_n:])

    m0 = 0.0 if x_right == x_left else (y_right - y_left) / (x_right - x_left)
    b0 = float(np.median(yfit) - m0 * np.median(xfit))

    baseline = m0 * xfit + b0
    A0 = float(np.max(yfit - baseline))
    if not np.isfinite(A0) or A0 <= 0:
        A0 = max(float(np.max(yfit) - np.min(yfit)), 1.0)

    sigma0 = max((xmax - xmin) / 8.0, 1.0)

    p0 = [A0, mu_guess, sigma0, m0, b0]
    bounds = (
        [0.0, xmin, 0.5, -np.inf, -np.inf],
        [np.inf, xmax, max(xmax - xmin, 2.0), np.inf, np.inf],
    )

    popt, pcov = curve_fit(
        gaussian_plus_linear,
        xfit,
        yfit,
        p0=p0,
        bounds=bounds,
        maxfev=50000,
    )

    perr = np.sqrt(np.diag(pcov)) if pcov is not None else np.full(len(popt), np.nan)

    resid = yfit - gaussian_plus_linear(xfit, *popt)
    dof = max(len(xfit) - len(popt), 1)
    chi2 = float(np.sum(resid ** 2))
    redchi2 = chi2 / dof

    A, mu, sigma, m, b = map(float, popt)
    A_err, mu_err, sigma_err, m_err, b_err = map(float, perr)

    xs_dense = np.linspace(xmin, xmax, 400)
    y_dense = gaussian_plus_linear(xs_dense, *popt)
    y_gauss = gaussian(xs_dense, A, mu, sigma)
    area = float(A * abs(sigma) * np.sqrt(2.0 * np.pi))

    return {
        "feature_label": feature_label,
        "fit_xmin": float(xmin),
        "fit_xmax": float(xmax),
        "center_ADC": mu,
        "center_ADC_err": mu_err,
        "sigma_ADC": abs(sigma),
        "sigma_ADC_err": sigma_err,
        "amplitude_counts_per_s": A,
        "amplitude_counts_per_s_err": A_err,
        "gaussian_area_counts": area,
        "gaussian_area_counts_err": area * np.sqrt(
            (A_err / A) ** 2 + (sigma_err / sigma) ** 2
        ) if A > 0 and sigma != 0 and np.isfinite(A_err) and np.isfinite(sigma_err) else np.nan,
        "background_slope": m,
        "background_slope_err": m_err,
        "background_intercept": b,
        "background_intercept_err": b_err,
        "fit_chi2": chi2,
        "fit_redchi2": redchi2,
        "xfit": xfit,
        "yfit": yfit,
        "xs_dense": xs_dense,
        "y_dense": y_dense,
        "y_gauss": y_gauss,
    }
```

```python
# -----------------------------
# FIT ONE FILE
# -----------------------------
def fit_compton_features_for_file(file_relpath, windows_row, make_plot=True):
    x_adc, raw_rate, bg_rate, adjusted_rate, meta_s, meta_b = background_subtracted_rate(file_relpath)

    back = fit_feature_window(
        x_adc, adjusted_rate,
        windows_row["backscatter_xmin"],
        windows_row["backscatter_xmax"],
        windows_row["backscatter_guess"],
        "backscatter_peak",
        file_relpath,
    )

    edge = fit_feature_window(
        x_adc, adjusted_rate,
        windows_row["edge_xmin"],
        windows_row["edge_xmax"],
        windows_row["edge_guess"],
        "compton_edge",
        file_relpath,
    )

    back_E, back_E_err, slope, slope_err, offset, offset_err, extra_sigma, day, preamp = adc_to_energy(
        back["center_ADC"], back["center_ADC_err"], file_relpath
    )
    edge_E, edge_E_err, _, _, _, _, _, _, _ = adc_to_energy(
        edge["center_ADC"], edge["center_ADC_err"], file_relpath
    )

    sigma_back_keV = abs(slope) * back["sigma_ADC"]
    sigma_back_keV_err = np.sqrt((back["sigma_ADC"] * slope_err) ** 2 + (slope * back["sigma_ADC_err"]) ** 2)

    sigma_edge_keV = abs(slope) * edge["sigma_ADC"]
    sigma_edge_keV_err = np.sqrt((edge["sigma_ADC"] * slope_err) ** 2 + (slope * edge["sigma_ADC_err"]) ** 2)

    length_adc = edge["center_ADC"] - back["center_ADC"]
    length_adc_err = np.sqrt(back["center_ADC_err"] ** 2 + edge["center_ADC_err"] ** 2)

    length_keV = edge_E - back_E
    length_keV_err = np.sqrt(back_E_err ** 2 + edge_E_err ** 2)

    region_mask = (x_adc >= back["center_ADC"]) & (x_adc <= edge["center_ADC"])
    x_region = np.asarray(x_adc[region_mask], dtype=float)
    y_region = np.asarray(adjusted_rate[region_mask], dtype=float)

    region_slope_adc = np.nan
    region_intercept = np.nan
    region_slope_adc_err = np.nan
    region_intercept_err = np.nan
    region_redchi2 = np.nan

    if len(x_region) >= 3:
        coeffs, cov = np.polyfit(x_region, y_region, deg=1, cov=True)
        region_slope_adc = float(coeffs[0])
        region_intercept = float(coeffs[1])
        region_slope_adc_err = float(np.sqrt(cov[0, 0]))
        region_intercept_err = float(np.sqrt(cov[1, 1]))

        y_model = region_slope_adc * x_region + region_intercept
        resid = y_region - y_model
        dof = max(len(x_region) - 2, 1)
        region_redchi2 = float(np.sum(resid ** 2) / dof)

    region_slope_keV = region_slope_adc / slope
    region_slope_keV_err = np.sqrt(
        (region_slope_adc_err / slope) ** 2 +
        ((region_slope_adc * slope_err) / (slope ** 2)) ** 2
    ) if slope != 0 else np.nan

    rows = []
    for fit_result, center_keV, center_keV_err, sigma_keV, sigma_keV_err in [
        (back, back_E, back_E_err, sigma_back_keV, sigma_back_keV_err),
        (edge, edge_E, edge_E_err, sigma_edge_keV, sigma_edge_keV_err),
    ]:
        row = {
            "file": file_relpath,
            "day": day,
            "preamp": preamp,
            "feature_label": fit_result["feature_label"],
            "center_ADC": fit_result["center_ADC"],
            "center_ADC_err": fit_result["center_ADC_err"],
            "center_keV": float(center_keV),
            "center_keV_err": float(center_keV_err),
            "sigma_ADC": fit_result["sigma_ADC"],
            "sigma_ADC_err": fit_result["sigma_ADC_err"],
            "sigma_keV": float(sigma_keV),
            "sigma_keV_err": float(sigma_keV_err),
            "amplitude_counts_per_s": fit_result["amplitude_counts_per_s"],
            "amplitude_counts_per_s_err": fit_result["amplitude_counts_per_s_err"],
            "gaussian_area_counts": fit_result["gaussian_area_counts"],
            "gaussian_area_counts_err": fit_result["gaussian_area_counts_err"],
            "fit_xmin": fit_result["fit_xmin"],
            "fit_xmax": fit_result["fit_xmax"],
            "background_slope": fit_result["background_slope"],
            "background_slope_err": fit_result["background_slope_err"],
            "background_intercept": fit_result["background_intercept"],
            "background_intercept_err": fit_result["background_intercept_err"],
            "fit_chi2": fit_result["fit_chi2"],
            "fit_redchi2": fit_result["fit_redchi2"],
            "calibration_slope_keV_per_ADC": slope,
            "calibration_slope_err": slope_err,
            "calibration_offset_keV": offset,
            "calibration_offset_err": offset_err,
            "calibration_extra_sigma_keV": extra_sigma,
            "compton_region_length_ADC": float(length_adc),
            "compton_region_length_ADC_err": float(length_adc_err),
            "compton_region_length_keV": float(length_keV),
            "compton_region_length_keV_err": float(length_keV_err),
            "compton_region_slope_counts_per_s_per_ADC": float(region_slope_adc),
            "compton_region_slope_err_per_ADC": float(region_slope_adc_err),
            "compton_region_slope_counts_per_s_per_keV": float(region_slope_keV),
            "compton_region_slope_err_per_keV": float(region_slope_keV_err),
            "compton_region_intercept": float(region_intercept),
            "compton_region_intercept_err": float(region_intercept_err),
            "compton_region_redchi2": float(region_redchi2),
            "compton_region_average_drop_per_ADC": float(region_slope_adc),
            "compton_region_average_drop_per_keV": float(region_slope_keV),
        }
        rows.append(row)

    if make_plot:
        adc_err_zeros = np.zeros_like(x_adc, dtype=float)
        x_keV, _, _, _, _, _, _, _, _ = adc_to_energy(x_adc, adc_err_zeros, file_relpath)

        fig, ax = plt.subplots(figsize=(12, 6))
        ax.plot(x_keV, adjusted_rate, lw=1.2, label="Background-subtracted spectrum")

        for fit_result, color in [(back, "tab:orange"), (edge, "tab:green")]:
            xs_keV, _, _, _, _, _, _, _, _ = adc_to_energy(
                fit_result["xs_dense"], np.zeros_like(fit_result["xs_dense"]), file_relpath
            )
            center_keV_val, _, _, _, _, _, _, _, _ = adc_to_energy(
                fit_result["center_ADC"], fit_result["center_ADC_err"], file_relpath
            )

            ax.axvspan(
                slope * fit_result["fit_xmin"] + offset,
                slope * fit_result["fit_xmax"] + offset,
                alpha=0.12,
                color=color,
            )
            ax.plot(xs_keV, fit_result["y_dense"], color=color, lw=2, label=fit_result["feature_label"])
            ax.axvline(float(center_keV_val), color=color, ls="--", lw=1.5)

        ax.errorbar(
            [float(back_E), float(edge_E)],
            [back["amplitude_counts_per_s"], edge["amplitude_counts_per_s"]],
            xerr=[float(back_E_err), float(edge_E_err)],
            yerr=[back["amplitude_counts_per_s_err"], edge["amplitude_counts_per_s_err"]],
            fmt="o",
            capsize=3,
            color="black",
            label="Fit centers ± errors",
        )

        ax.set_xlabel("Energy (keV)")
        ax.set_ylabel("Background-subtracted rate (counts/s)")
        ax.set_title(f"{file_relpath}\nBackscatter and Compton-edge proxy fits")
        ax.grid(True, alpha=0.3)
        ax.legend()

        plot_path = PLOT_DIR / f"{Path(file_relpath).stem}_compton_fits.png"
        fig.savefig(plot_path, dpi=200, bbox_inches="tight")
        plt.show()

    return pd.DataFrame(rows)
```

```python
# -----------------------------
# RUN FITS FOR ALL FILLED ROWS
# -----------------------------
if len(manual_windows) == 0:
    print("manual_windows is empty. Fill it in, then rerun this cell.")
else:
    all_rows = []
    failed = []

    for _, row in manual_windows.iterrows():
        file_relpath = row["file"]
        try:
            result_df = fit_compton_features_for_file(file_relpath, row, make_plot=True)
            all_rows.append(result_df)

            per_run_path = PER_RUN_DIR / f"{Path(file_relpath).stem}_compton_features.csv"
            result_df.to_csv(per_run_path, index=False)
            print("Saved:", per_run_path)
        except Exception as exc:
            failed.append({"file": file_relpath, "error": repr(exc)})
            print("FAILED:", file_relpath)
            print("   ", repr(exc))

    if len(all_rows) > 0:
        compton_master = pd.concat(all_rows, ignore_index=True)
        compton_master.to_csv(MASTER_FILE, index=False)
        print("\nSaved master file:", MASTER_FILE)
        display(compton_master.head(20))
    else:
        compton_master = pd.DataFrame()
        print("No successful fits.")

    failed_df = pd.DataFrame(failed)
    if len(failed_df) > 0:
        display(failed_df)
```
