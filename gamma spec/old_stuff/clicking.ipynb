```python
# -----------------------------
# CELL 1: FORCE A GUI BACKEND
# -----------------------------
# Run this FIRST in a fresh kernel.

import matplotlib

_backend_errors = []

for candidate in ["QtAgg", "TkAgg"]:
    try:
        matplotlib.use(candidate, force=True)
        print(f"Using backend: {candidate}")
        break
    except Exception as e:
        _backend_errors.append((candidate, repr(e)))
else:
    raise RuntimeError(
        "Could not start a GUI matplotlib backend.\n"
        f"Tried: {_backend_errors}\n"
        "Install one of the GUI toolkits, e.g.\n"
        "  conda install pyqt\n"
        "or\n"
        "  conda install tk"
    )
```

```python
# -----------------------------
# CELL 2: IMPORTS
# -----------------------------

import sys
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

print("Python:", sys.version)
print("matplotlib:", matplotlib.__version__)
print("backend:", matplotlib.get_backend())
```

```python
# -----------------------------
# CELL 3: SYNTHETIC DEBUG DATA
# -----------------------------

np.random.seed(7)

x = np.linspace(0, 1000, 2000)

def gauss(x, A, mu, sigma):
    return A * np.exp(-(x - mu)**2 / (2 * sigma**2))

y = (
    gauss(x, 120, 180, 18)
    + gauss(x, 200, 430, 25)
    + gauss(x, 150, 760, 20)
    + 12
    + 0.01 * x
    + np.random.normal(0, 3, size=len(x))
)

print("Synthetic data ready.")
```

```python
# -----------------------------
# CELL 4: CLICK STATE
# -----------------------------

CLICKS = []

def nearest_index(xarr, xclick):
    return int(np.argmin(np.abs(xarr - xclick)))

print("Click storage initialized.")
```

```python
# -----------------------------
# CELL 5: LAUNCH CLICK WINDOW
# -----------------------------

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, lw=1)
ax.set_title("Debug click window")
ax.set_xlabel("Channel")
ax.set_ylabel("Counts")
ax.grid(True, alpha=0.3)

status_text = ax.text(
    0.02, 0.98, "Clicks: 0",
    transform=ax.transAxes,
    va="top"
)

def on_click(event):
    if event.inaxes != ax or event.xdata is None:
        return

    idx = nearest_index(x, event.xdata)
    xs = float(x[idx])
    ys = float(y[idx])

    CLICKS.append((xs, ys, idx))
    ax.plot(xs, ys, marker="o", ms=8)
    status_text.set_text(
        f"Clicks: {len(CLICKS)}\nLast: x={xs:.3f}, y={ys:.3f}, idx={idx}"
    )
    fig.canvas.draw_idle()

    print(f"captured click #{len(CLICKS)}: x={xs:.6f}, y={ys:.6f}, idx={idx}")

cid = fig.canvas.mpl_connect("button_press_event", on_click)

print("A separate plot window should open.")
print("Click directly on the plot.")
print("When done, close the window and run the next cell.")

plt.show()
```

```python
# -----------------------------
# CELL 6: INSPECT CAPTURED CLICKS
# -----------------------------

print("Captured clicks:")
for i, (xs, ys, idx) in enumerate(CLICKS, start=1):
    print(f"{i:2d}: x={xs:.6f}, y={ys:.6f}, idx={idx}")

print("\nAs array:")
print(np.array(CLICKS, dtype=float) if CLICKS else "[]")
```

```python
# -----------------------------
# CELL 7: CLEAR CLICKS
# -----------------------------

CLICKS.clear()
print("Clicks cleared.")
```
