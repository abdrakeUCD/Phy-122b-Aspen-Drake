```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print("backend =", matplotlib.get_backend())
```

```python
clicked_points = []

x = np.linspace(0, 10, 500)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Click inside the axes")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
```

```python
def onclick(event):
    global clicked_points

    if event.inaxes != ax:
        return

    clicked_points.append((event.xdata, event.ydata))

    # visual proof that callback fired
    ax.axvline(event.xdata, linestyle="--")
    ax.plot(event.xdata, event.ydata, marker="o")
    ax.set_title(f"Last click: x={event.xdata:.3f}, y={event.ydata:.3f}")
    fig.canvas.draw_idle()

cid = fig.canvas.mpl_connect("button_press_event", onclick)

fig
```
